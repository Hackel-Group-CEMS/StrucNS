```python
import matplotlib.pyplot as plt

# Data
labels = ['Connectivity-Based', 'Projection Based', 'Geometry Based']
sizes = [62.120495, 28.002691, 9.876813]
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

FONT_SIZE = 20

fig, ax = plt.subplots(figsize=(12, 12))

# labeldistance: 0.6 moves the text labels inside the circle
# pctdistance: 0.85 places the percentages further out but still inside
wedges, texts, autotexts = ax.pie(
    sizes, 
    labels=labels, 
    autopct='%1.1f%%', 
    startangle=140, 
    colors=colors,
    labeldistance=0.6,  
    pctdistance=0.85,    
    textprops={'fontsize': FONT_SIZE, 'fontweight': 'bold'}
)

# Center-align the text labels for a cleaner look inside the slices
for text in texts:
    text.set_horizontalalignment('center')

ax.set_title('Importance Share (%) by Category', fontsize=FONT_SIZE + 8, pad=30)

plt.axis('equal')
plt.show()
```
